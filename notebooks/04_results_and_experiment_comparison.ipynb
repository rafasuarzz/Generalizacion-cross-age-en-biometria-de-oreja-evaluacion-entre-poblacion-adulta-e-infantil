This notebook visualizes training curves and compares the results across experiments using metrics such as ROC-AUC and EER.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("all_experiments_results.csv")

df["experiment"] = df["train_datasets"] + " -> " + df["test_datasets"]

# ROC-AUC
plt.figure(figsize=(10, 5))
plt.bar(df["experiment"], df["roc_auc"])
plt.xticks(rotation=45, ha="right")
plt.ylabel("ROC-AUC")
plt.title("Comparación de ROC-AUC por experimento")
plt.tight_layout()
plt.savefig("compare_auc.png")
plt.show()

# EER
plt.figure(figsize=(10, 5))
plt.bar(df["experiment"], df["eer"])
plt.xticks(rotation=45, ha="right")
plt.ylabel("EER")
plt.title("Comparación de EER por experimento")
plt.tight_layout()
plt.savefig("compare_eer.png")
plt.show()

# Rank-1
plt.figure(figsize=(10, 5))
plt.bar(df["experiment"], df["rank1"])
plt.xticks(rotation=45, ha="right")
plt.ylabel("Rank-1")
plt.title("Comparación de Rank-1 por experimento")
plt.tight_layout()
plt.savefig("compare_rank1.png")
plt.show()
```
